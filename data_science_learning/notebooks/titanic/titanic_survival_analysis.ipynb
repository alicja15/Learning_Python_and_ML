# Titanic Survival Analysis

Exploratory data analysis of the Titanic passenger dataset (loaded from Seaborn's built-in datasets).
One feature-engineering step (family size) and a 2x2 dashboard of survival patterns.

**Questions explored**
1. How do passenger class and sex affect survival?
2. How does the age distribution differ between survivors and victims?
3. Is there a "family size paradox" (is it better to travel alone, in a small family, or in a big one)?
4. Which numeric variables are correlated?

> `sns.load_dataset("titanic")` downloads the data on first use, so an internet connection is required.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

df = sns.load_dataset("titanic")

# Feature engineering: family size
df["family_size"] = df["sibsp"] + df["parch"] + 1
df["family_type"] = pd.cut(
    df["family_size"],
    bins=[0, 1, 4, 11],
    labels=["Alone (1)", "Small family (2-4)", "Big family (5+)"],
)

# Plot aesthetics
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(15, 11))

# Plot 1: survival by class and sex
sns.barplot(
    data=df,
    x="class",
    y="survived",
    hue="sex",
    palette="Blues_d",
    ax=axes[0, 0],
    errorbar=None,
)
axes[0, 0].set_title("1. Effect of Class and Sex on Survival", fontsize=12)
axes[0, 0].set_ylabel("Survival rate")
axes[0, 0].set_ylim(0, 1)

# Plot 2: age distribution by group (who) and survival
sns.violinplot(
    data=df,
    x="who",
    y="age",
    hue="survived",
    split=True,
    inner="quart",
    palette="Set2",
    ax=axes[0, 1],
)
axes[0, 1].set_title("2. Age Distribution of Survivors and Victims", fontsize=12)

# Plot 3: the family size paradox
sns.barplot(
    data=df,
    x="family_type",
    y="survived",
    hue="class",
    palette="viridis",
    ax=axes[1, 0],
    errorbar=None,
)
axes[1, 0].set_title("3. Survival by Family Size and Class", fontsize=12)
axes[1, 0].set_ylabel("Survival rate")

# Plot 4: correlation matrix of numeric features
numeric_cols = df.select_dtypes(include=[np.number]).columns
sns.heatmap(
    df[numeric_cols].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    ax=axes[1, 1],
)
axes[1, 1].set_title("4. Correlations of Numeric Variables", fontsize=12)

plt.tight_layout()
plt.show()
```
